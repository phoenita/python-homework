## Задание 1. Средние буквы слова

Дано непустое слово из латинских букв. Если длина нечётная, выводится одна средняя буква, если чётная, две средние.

`len(word)` определяет длину слова, `// 2` вычисляет индекс середины, `% 2` проверяет чётность. Индексы начинаются с нуля; правая граница среза не включается.

### Пример 1: test

```python
word = 'test'

middle = len(word) // 2

if len(word) % 2 == 0:
    print(word[middle - 1:middle + 1])
else:
    print(word[middle])
```

### Пример 2: testing

```python
word = 'testing'

middle = len(word) // 2

if len(word) % 2 == 0:
    print(word[middle - 1:middle + 1])
else:
    print(word[middle])
```

## Задание 2. Подбор пар

Сначала сравниваем длины списков. Если они различаются, выводим предупреждение и не формируем пары. При равных длинах сортируем оба списка и объединяем имена с одинаковыми индексами.

`sorted` возвращает отсортированный список, сохраняя исходный. `zip` объединяет соответствующие элементы двух списков, а цикл `for` выводит пары.

### Пример 1: равное количество участников

```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

if len(boys) != len(girls):
    print('Внимание, кто-то может остаться без пары!')
else:
    print('Идеальные пары:')
    for boy, girl in zip(sorted(boys), sorted(girls)):
        print(boy, 'и', girl)
```

### Пример 2: разное количество участников

```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard', 'Michael']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

if len(boys) != len(girls):
    print('Внимание, кто-то может остаться без пары!')
else:
    print('Идеальные пары:')
    for boy, girl in zip(sorted(boys), sorted(girls)):
        print(boy, 'и', girl)
```
