## Задание 1. Средние буквы слова

Я вычисляю середину непустого слова через `len(word) // 2`. При нечётной длине вывожу одну букву, при чётной использую срез для двух средних букв.

### Пример 1: test

```python
word = 'test'

middle = len(word) // 2

if len(word) % 2 == 0:
    print(word[middle - 1:middle + 1])
else:
    print(word[middle])
```

### Пример 2: testing

```python
word = 'testing'

middle = len(word) // 2

if len(word) % 2 == 0:
    print(word[middle - 1:middle + 1])
else:
    print(word[middle])
```

## Задание 2. Подбор пар

Я сравниваю длины списков. Если они различаются, вывожу предупреждение. При равных длинах сортирую списки через `sorted` и объединяю имена с одинаковыми индексами через `zip`. Пары вывожу в цикле `for`.

### Пример 1: равное количество участников

```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

if len(boys) != len(girls):
    print('Внимание, кто-то может остаться без пары!')
else:
    print('Идеальные пары:')
    for boy, girl in zip(sorted(boys), sorted(girls)):
        print(boy, 'и', girl)
```

### Пример 2: разное количество участников

```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard', 'Michael']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

if len(boys) != len(girls):
    print('Внимание, кто-то может остаться без пары!')
else:
    print('Идеальные пары:')
    for boy, girl in zip(sorted(boys), sorted(girls)):
        print(boy, 'и', girl)
```
