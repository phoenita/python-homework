## Задание 1. Високосный год

```python
year = 2024

# Високосный год делится на 400 или делится на 4, но не на 100.
if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("Високосный год")
else:
    print("Обычный год")
```

## Задание 2. Счастливый билет

```python
number = 123456

# Выделяю цифры с помощью целочисленного деления и остатка от деления.
first_sum = number // 100000 + number // 10000 % 10 + number // 1000 % 10
last_sum = number // 100 % 10 + number // 10 % 10 + number % 10

if first_sum == last_sum:
    print("Счастливый билет")
else:
    print("Несчастливый билет")
```
