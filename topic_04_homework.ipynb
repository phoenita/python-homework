Выполнены обязательное и дополнительное задания. Программа на Python 3 ищет владельца документа и полку по номеру. Дополнительные библиотеки не нужны.

## 1. Исходные данные

Номера документов и полок хранятся как строки. Единственные глобальные переменные с данными: `documents` и `directories`.

```python
documents = [
    {'type': 'passport', 'number': '2207 876234', 'name': 'Василий Гупкин'},
    {'type': 'invoice', 'number': '11-2', 'name': 'Геннадий Покемонов'},
    {'type': 'insurance', 'number': '10006', 'name': 'Аристарх Павлов'}
]

directories = {
    '1': ['2207 876234', '11-2'],
    '2': ['10006'],
    '3': []
}
```

## 2. Функции

`find_owner` ищет владельца, `find_shelf` ищет полку, `main` отвечает за ввод команд и вывод результата. Данные передаются в функции через параметры. Цикл `while` работает до команды `q`. Неизвестная команда не завершает программу.

```python
def find_owner(documents, number):
    """Возвращает владельца документа или сообщение об отсутствии владельца."""
    for document in documents:
        if document['number'] == number:
            return document['name']
    return 'владелец не найден'


def find_shelf(directories, number):
    """Возвращает номер полки или None, если документ не найден."""
    for shelf, numbers in directories.items():
        if number in numbers:
            return shelf
    return None


def main(documents, directories):
    """Обрабатывает команды пользователя до ввода q."""
    while True:
        command = input('Введите команду: ').strip().lower()

        if command == 'q':
            print('Программа завершена.')
            break

        if command not in ('p', 's'):
            print('Неизвестная команда. Используйте p, s или q.')
            continue

        number = input('Введите номер документа: ').strip()

        if command == 'p':
            print('Владелец документа:', find_owner(documents, number))
        else:
            shelf = find_shelf(directories, number)
            if shelf is None:
                print('Документ не найден.')
            else:
                print('Документ хранится на полке:', shelf)
```

## 3. Самопроверка

Проверяются все примеры из чек-листа, остальные документы, неизвестная команда и выход. Подмена ввода используется только в проверках; вспомогательные переменные и импорты находятся внутри функции.

```python
def self_check(documents, directories):
    """Проверяет поиск и команды с заранее заданным вводом."""
    from contextlib import redirect_stdout
    from io import StringIO
    from unittest.mock import patch

    assert find_owner(documents, '10006') == 'Аристарх Павлов'
    assert find_owner(documents, '1001') == 'владелец не найден'
    assert find_shelf(directories, '10006') == '2'
    assert find_shelf(directories, '1001') is None

    # Все документы присутствуют на полках; поиск работает для каждого из них.
    for document in documents:
        assert find_owner(documents, document['number']) == document['name']
        assert find_shelf(directories, document['number']) is not None

    # Один сеанс включает все примеры, неизвестную команду и выход.
    answers = ['p', '10006', 'p', '1001', 's', '10006', 's', '1001', 'x', 'q']
    output = StringIO()
    with patch('builtins.input', side_effect=answers) as mocked_input:
        with redirect_stdout(output):
            main(documents, directories)
    assert mocked_input.call_count == len(answers)
    assert output.getvalue().splitlines() == [
        'Владелец документа: Аристарх Павлов',
        'Владелец документа: владелец не найден',
        'Документ хранится на полке: 2',
        'Документ не найден.',
        'Неизвестная команда. Используйте p, s или q.',
        'Программа завершена.'
    ]

    # Команда q завершает программу без запроса номера документа.
    with patch('builtins.input', side_effect=['q']) as mocked_input:
        with redirect_stdout(StringIO()):
            main(documents, directories)
    assert mocked_input.call_count == 1
    print('Все проверки пройдены.')


self_check(documents, directories)
```
